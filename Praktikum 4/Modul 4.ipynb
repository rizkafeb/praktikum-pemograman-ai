```python
# LATIHAN 1

N = int(input("Masukkan jumlah jenis barang (N): "))
total_belanja = 0

for i in range(1, N+1):
    print(f"\nBarang ke-{i}")
    harga = int(input(f"  Harga barang ke-{i}: Rp "))
    jumlah = int(input(f"  Jumlah barang ke-{i}: "))
    subtotal = harga * jumlah
    print(f"  Subtotal: Rp {subtotal}")
    total_belanja = total_belanja + subtotal  # atau total_belanja += subtotal

print(f"\nTotal belanja {N} jenis barang adalah: Rp {total_belanja}")
```

```python
#LATIHAN 2

N = int(input("Masukkan banyaknya bilangan (N): "))

genap = 0
ganjil = 0
daftar_genap = []
daftar_ganjil = []


for i in range(N):
    angka = int(input(f"Masukkan bilangan ke-{i+1}: "))
    if angka % 2 == 0:
        genap += 1
        daftar_genap.append(angka)
    else:
        ganjil += 1
        daftar_ganjil.append(angka)

print("\n=== HASIL ===")
print(f"Bilangan genap  : {daftar_genap}")
print(f"Jumlah genap    : {genap}")
print(f"Bilangan ganjil : {daftar_ganjil}")
print(f"Jumlah ganjil   : {ganjil}")
```

```python
#LATIHAN 5

n = int(input("Masukkan jumlah mahasiswa: "))

data_mahasiswa = []

for i in range(n):
    print(f"\nData Mahasiswa ke-{i + 1}")
    nama = input("Nama Mahasiswa : ")
    tugas = float(input("Nilai Tugas    : "))
    kuis = float(input("Nilai Kuis     : "))
    uts = float(input("Nilai UTS      : "))
    uas = float(input("Nilai UAS      : "))
    
    # Hitung nilai akhir (Rata-rata)
    nilai_akhir = (tugas + kuis + uts + uas) / 4
    
    data_mahasiswa.append({
        "nama": nama,
        "tugas": tugas,
        "kuis": kuis,
        "uts": uts,
        "uas": uas,
        "akhir": nilai_akhir
    })

# Menampilkan Tabel Output
print("\n" + "=" * 65)
print(f"{'No.':<4} {'Nama Mhs':<15} {'N.Tugas':<10} {'N.Kuis':<8} {'N.UTS':<8} {'N.UAS':<8} {'NilaiAkhir':<10}")
print("=" * 65)

for idx, mhs in enumerate(data_mahasiswa, start=1):
    print(f"{idx:<4} {mhs['nama']:<15} {mhs['tugas']:<10.1f} {mhs['kuis']:<8.1f} {mhs['uts']:<8.1f} {mhs['uas']:<8.1f} {mhs['akhir']:<10.2f}")

print("=" * 65)
```
