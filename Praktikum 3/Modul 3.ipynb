```python
#Pernyataan berkondisi, menentukan Diskriminan dari Persamaan Kuadrat (gambar 3.2)

a=3
b=5
c=1
D=b*b-4*a*c

if (D<0):
    print("akar Imaginer")
elif (D>0): 
    print("akar nyata berlainan")
else: 
    print("akar kembar")
```

```python
#Pernyataan berkondisi, menentukan Kriteria Umur (gambar 3.3)

umur = int(input("Umur Anda: "))

if umur <= 5:
    kriteria = "Balita"
elif umur <= 13:
    kriteria = "Anak-anak"
elif umur <= 25:
    kriteria = "Remaja"
elif umur <= 35:
    kriteria = "Dewasa"
elif umur <= 55:
    kriteria = "Orang Tua"
else:
    kriteria = "Lansia"

print("Umur %d dan kriteria yaitu %s" % (umur, kriteria))
```

```python
#Pernyataan if bersarang (gambar 3.4)

Nama = input('Nama Karyawan: ')
NIK = input('NIK Karyawan: ')
Status = input('Status (Staff/Honor): ')
Gol = input('Golongan (A/B/C): ')

if Status == 'Staff':
    Gaji = 1000000

    if Gol == 'A':
        Bonus = 200000
    elif Gol == 'B':
        Bonus = 400000
    elif Gol == 'C':
        Bonus = 550000
    else:
        Bonus = 0

elif Status == 'Honor':
    Gaji = 750000

    if Gol == 'A':
        Bonus = 150000
    elif Gol == 'B':
        Bonus = 275000
    elif Gol == 'C':
        Bonus = 480000
    else:
        Bonus = 0

else:
    Gaji = 0
    Bonus = 0
    print('Status karyawan tidak valid')

Gatot = Gaji + Bonus

print('Nama Karyawan:', Nama)
print('NIK Karyawan:', NIK)
print('Gaji Rp', Gaji)
print('Bonus Rp', Bonus)
print('Gaji Total Rp', Gatot)
```
