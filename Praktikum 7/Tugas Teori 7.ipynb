```python

import random

print("=== PERMAINAN GUNTING BATU KERTAS ===")

pilihan = ["gunting", "batu", "kertas"]

while True:
    pemain = input("\nPilih gunting, batu, atau kertas: ").lower()

    if pemain not in pilihan:
        print("Pilihan tidak tersedia!")
        continue

    komputer = random.choice(pilihan)

    print("Pilihan kamu    :", pemain)
    print("Pilihan komputer:", komputer)

    if pemain == komputer:
        print("Hasil: SERI!")
    elif (pemain == "gunting" and komputer == "kertas") or \
         (pemain == "batu" and komputer == "gunting") or \
         (pemain == "kertas" and komputer == "batu"):
        print("Hasil: KAMU MENANG!")
    else:
        print("Hasil: KAMU KALAH!")

    ulang = input("\nMain lagi? (y/n): ").lower()

    if ulang != "y":
        print("Terima kasih sudah bermain!")
        break
```

```python
inputan = input("Inputan: ")

angka = 0
huruf = 0

for x in inputan:
    if x.isdigit():
        angka = angka + 1
    elif x.isalpha():
        huruf = huruf + 1

print("Jumlah Angka :", angka)
print("Jumlah Huruf :", huruf)
```
